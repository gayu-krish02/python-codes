```python
# Air-drawing with your index finger using OpenCV and MediaPipe with Tools and White Background
import cv2
import mediapipe as mp
import numpy as np

# Initialize MediaPipe Hands
mp_hands = mp.solutions.hands
hands = mp_hands.Hands(max_num_hands=1, min_detection_confidence=0.7)
mp_draw = mp.solutions.drawing_utils

cap = cv2.VideoCapture(0)
canvas = None
prev_point = None

# Define color and thickness options
colors = [(0, 0, 255), (0, 255, 0), (255, 0, 0)]  # Red, Green, Blue
color_names = ['Red', 'Green', 'Blue']
color_index = 0

# Tool settings
tools = ['Brush', 'Pencil', 'Sketch']
tool_index = 0
tool_thickness = [12, 2, 6]  # Brush = thickmost, Pencil = light/thin, Sketch = moderate
tool_color_intensity = [1.0, 0.3, 0.6]  # Brush = full color, Pencil = faded, Sketch = mid

# Define button positions
button_y = 10
color_buttons = [(10 + i*70, button_y, 60, 40) for i in range(len(colors))]
tool_buttons = [(10 + i*100, 60, 90, 40) for i in range(len(tools))]
erase_button = (520, 110, 120, 40)  # Erase on right side

def draw_buttons(img):
    for i, (x, y, w, h) in enumerate(color_buttons):
        cv2.rectangle(img, (x, y), (x+w, y+h), colors[i], -1)
        cv2.putText(img, color_names[i], (x+5, y+25), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (255,255,255), 2)

    for i, (x, y, w, h) in enumerate(tool_buttons):
        cv2.rectangle(img, (x, y), (x+w, y+h), (50,50,50), -1)
        cv2.putText(img, tools[i], (x+5, y+25), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (255,255,255), 2)

    x, y, w, h = erase_button
    cv2.rectangle(img, (x, y), (x+w, y+h), (0, 0, 0), -1)
    cv2.putText(img, 'Erase', (x+10, y+25), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (255,255,255), 2)

while True:
    ret, frame = cap.read()
    if not ret:
        break

    frame = cv2.flip(frame, 1)
    
    # Initialize canvas with white background instead of black
    if canvas is None:
        canvas = np.ones_like(frame) * 255  # White background

    rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    results = hands.process(rgb)
    index_finger_tip = None
    index_finger_up = False

    # Create a separate image to draw buttons (on white background)
    button_layer = np.ones_like(frame) * 255
    draw_buttons(button_layer)

    if results.multi_hand_landmarks:
        for hand_landmarks in results.multi_hand_landmarks:
            h, w, _ = frame.shape
            landmarks = hand_landmarks.landmark
            x = int(landmarks[8].x * w)
            y = int(landmarks[8].y * h)
            index_finger_tip = (x, y)

            # Check if index finger is up (tip above second joint)
            if landmarks[8].y < landmarks[6].y:
                index_finger_up = True

            mp_draw.draw_landmarks(button_layer, hand_landmarks, mp_hands.HAND_CONNECTIONS)

    if index_finger_tip:
        x, y = index_finger_tip

        # Check for color button click
        for i, (bx, by, bw, bh) in enumerate(color_buttons):
            if bx < x < bx + bw and by < y < by + bh:
                color_index = i

        # Check for tool button click
        for i, (bx, by, bw, bh) in enumerate(tool_buttons):
            if bx < x < bx + bw and by < y < by + bh:
                tool_index = i

        # Check for erase button click
        ex, ey, ew, eh = erase_button
        if ex < x < ex + ew and ey < y < ey + eh:
            canvas = np.ones_like(frame) * 255  # Reset to white

        # Draw only if index finger up and not touching buttons (below button area)
        if index_finger_up and y > 160:
            if prev_point:
                thickness = tool_thickness[tool_index]
                base_color = np.array(colors[color_index], dtype=np.float32)
                faded_color = (base_color * tool_color_intensity[tool_index]).astype(np.uint8)
                faded_color = tuple(map(int, faded_color))
                cv2.line(canvas, prev_point, index_finger_tip, faded_color, thickness)
            prev_point = index_finger_tip
        else:
            prev_point = None

    # Overlay the buttons on top of the canvas
    output = canvas.copy()
    # Buttons drawn on white background, so just overlay
    mask = np.any(button_layer != 255, axis=2)
    output[mask] = button_layer[mask]

    cv2.imshow("Air Drawing - Index Finger", output)

    key = cv2.waitKey(1) & 0xFF
    if key == ord('c'):
        canvas = np.ones_like(frame) * 255  # Clear with white background
    if cv2.getWindowProperty("Air Drawing - Index Finger", cv2.WND_PROP_VISIBLE) < 1:
        break

cap.release()
cv2.destroyAllWindows()
```
